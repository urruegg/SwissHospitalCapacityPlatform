```python
# v3 change: DefaultAzureCredential does not work in Fabric Spark notebooks (no MI, no env vars, no CLI on path).
# Use notebookutils.credentials.getToken which brokers the current user's Entra token via Fabric.
import time
import datetime as _dt
from azure.cosmos import CosmosClient
from azure.core.credentials import AccessToken

COSMOS_URL = 'https://cosmos-csa-ihzhhpf-sit.documents.azure.com:443/'
DB_NAME = 'csa'
COSMOS_SCOPE = 'https://cosmos.azure.com/.default'

class FabricNBUToken:
    """Adapter that returns Fabric-brokered tokens in the shape azure.core expects."""
    def get_token(self, *scopes, **kwargs):
        # notebookutils.credentials.getToken accepts the target audience.
        # We ignore the requested scopes and always ask for the Cosmos audience.
        raw = notebookutils.credentials.getToken('https://cosmos.azure.com')
        # Tokens returned by Fabric are typically ~60 min; we conservatively cache for 55.
        return AccessToken(raw, int(time.time()) + 3300)

print('Requesting Cosmos token via notebookutils.credentials.getToken...')
cred = FabricNBUToken()
probe = cred.get_token(COSMOS_SCOPE)
print(f'Token acquired (expires in ~{(probe.expires_on - int(time.time()))//60} min)')

print('Connecting to Cosmos via Fabric MPE...')
client = CosmosClient(COSMOS_URL, credential=cred)
db = client.get_database_client(DB_NAME)

print('Containers reachable via MPE:')
containers = list(db.list_containers())
for c in containers:
    print(f'  - {c["id"]}')
print()
assert {c['id'] for c in containers} == {'scenarios', 'agent-memory', 'response-levers', 'simulation-runs'}, 'expected 4 containers'
print('OK - all 4 containers reachable via private endpoint')
```

```python
RULES_VERSION = '1.0.0'
TIER2_UTILIZATION_THRESHOLD = 0.90
DEFAULT_LEVER_HEADROOM = 0.15
SPECIAL_CAPABILITIES = frozenset({'burn-beds', 'ventilators', 'decontamination', 'isolation-beds'})
_EQUIPMENT_RESOURCES = frozenset({'ventilators', 'supplies', 'ppe', 'oxygen', 'blood'})

def _bump(o, u, v): return round(o * (1.0 + v / 100.0)) if u == 'percent' else round(o + v)
def _drop(c, u, v): return round(c * (1.0 - v / 100.0)) if u == 'percent' else round(c - v)

def apply_shock(baseline, shock):
    vector = shock.get('shockVector', 'demand-surge')
    affected = set(shock.get('affectedResources', []))
    mag = shock.get('magnitude', {}) or {}
    unit = mag.get('unit', 'percent'); value = float(mag.get('value', 0))
    projected = {}
    for r, dim in baseline.items():
        cap = int(dim.get('capacity', 0)); occ = int(dim.get('occupied', 0))
        if r in affected:
            if vector == 'demand-surge': occ = _bump(occ, unit, value)
            elif vector in ('capacity-loss', 'staff-loss'): cap = _drop(cap, unit, value)
            elif vector in ('supply-loss', 'system-loss'):
                if r in _EQUIPMENT_RESOURCES or vector == 'system-loss': cap = _drop(cap, unit, value)
        projected[r] = {'capacity': max(cap, 0), 'occupied': max(occ, 0)}
    return projected

def project_state(baseline, shock, lever_headroom=DEFAULT_LEVER_HEADROOM, flags=None):
    projected = apply_shock(baseline, shock); resources = {}; exceeded = False
    for name, dim in projected.items():
        c, o = dim['capacity'], dim['occupied']
        u = (o / c) if c > 0 else float('inf')
        s = max(o - c, 0)
        resources[name] = {'utilization': round(u, 4), 'shortfall': s}
        if o > c * (1.0 + lever_headroom): exceeded = True
    out_flags = dict(flags or {}); out_flags.setdefault('capacityExceededAfterLevers', exceeded)
    return {'resources': resources, 'flags': out_flags}

def summarize_kpis(state):
    rs = state.get('resources', {})
    total = sum(int(d.get('shortfall', 0)) for d in rs.values())
    us = [float(d.get('utilization', 0.0)) for d in rs.values() if d.get('utilization') not in (None, float('inf'))]
    peak = max(us) if us else 0.0
    return {'peakUtilization': round(peak, 4), 'totalShortfall': float(total), 'resourcesOverThreshold': float(sum(1 for u in us if u >= 0.90))}

def classify_tier(state):
    rs = state.get('resources', {}) or {}; fl = state.get('flags', {}) or {}; reasons = []
    if fl.get('capacityExceededAfterLevers'): reasons.append('demand exceeds site capacity even after internal levers')
    if fl.get('multiCanton'): reasons.append('multi-canton event (VKSD Art. 2)')
    if fl.get('severeConsequence'): reasons.append('severe-consequence event (VKSD Art. 2)')
    for name, dim in rs.items():
        u = float(dim.get('utilization', 0.0)); s = int(dim.get('shortfall', 0))
        if name in SPECIAL_CAPABILITIES and (u > 1.0 or s > 0): reasons.append(f'special capability overwhelmed: {name}')
    if reasons: return {'tier': 3, 'reasons': reasons, 'rulesVersion': RULES_VERSION}
    for name, dim in rs.items():
        u = float(dim.get('utilization', 0.0))
        if u >= TIER2_UTILIZATION_THRESHOLD:
            reasons.append(f"{name} utilization {u:.0%} breaches threshold {TIER2_UTILIZATION_THRESHOLD:.0%} - internal reallocation required")
    if reasons: return {'tier': 2, 'reasons': reasons, 'rulesVersion': RULES_VERSION}
    return {'tier': 1, 'reasons': ['all resource dimensions within normal operating capacity'], 'rulesVersion': RULES_VERSION}

print('Shock model + tier classifier loaded, rules v', RULES_VERSION)
```

```python
# Seed a minimum-viable subset (proves write path via MPE).
levers = db.get_container_client('response-levers')
scenarios = db.get_container_client('scenarios')

seed_levers = [
    {'id': 'lev-l2-open-surge-beds', 'leverId': 'lev-l2-open-surge-beds', 'tier': 2, 'family': 'internal-reallocation', 'label': 'Open surge beds', 'description': 'Activate pre-designated surge beds in step-down units.', 'doctrineRef': 'Swiss Lage doctrine (VKSD); ADR-0024 tier classifier'},
    {'id': 'lev-l2-delay-elective', 'leverId': 'lev-l2-delay-elective', 'tier': 2, 'family': 'internal-reallocation', 'label': 'Delay elective admissions', 'description': 'Defer non-urgent elective admissions by 24-72h.', 'doctrineRef': 'ADR-0024'},
    {'id': 'lev-l3-transfer-canton', 'leverId': 'lev-l3-transfer-canton', 'tier': 3, 'family': 'multi-canton-coordination', 'label': 'Inter-cantonal transfers', 'description': 'Coordinate with KSD/IES for cross-canton transfers.', 'doctrineRef': 'VKSD Art. 2; ADR-0024'},
]
for l in seed_levers: levers.upsert_item(l)

seed_scenarios = [{
    'id': 'pediatric-virus-surge-rsv', 'scenarioId': 'pediatric-virus-surge-rsv',
    'family': 'F6', 'title': 'Pediatric virus surge (RSV)',
    'trigger': 'seasonal', 'shockVector': 'demand-surge',
    'affectedResources': ['pediatric-beds'],
    'magnitude': {'unit': 'percent', 'value': 50},
    'onset': 'gradual', 'duration': 'weeks',
    'defaultTier': 2, 'mvpRequired': True,
    'responseLevers': ['lev-l2-open-surge-beds', 'lev-l2-delay-elective'],
}]
for s in seed_scenarios: scenarios.upsert_item(s)

print(f'Seeded {len(seed_levers)} levers + {len(seed_scenarios)} scenarios into Cosmos via MPE')
print('response-levers count:', sum(1 for _ in levers.read_all_items()))
print('scenarios count:', sum(1 for _ in scenarios.read_all_items()))
```

```python
# Run the canonical RSV simulation - same logic as csa-simulate.simulate().
import uuid

baseline = {'pediatric-beds': {'capacity': 40, 'occupied': 30}}
scenario = {
    'scenarioId': 'pediatric-virus-surge-rsv',
    'shockVector': 'demand-surge',
    'affectedResources': ['pediatric-beds'],
    'magnitude': {'unit': 'percent', 'value': 50}
}
run_id = f'run-rsv-live-{uuid.uuid4().hex[:8]}'

state = project_state(baseline, scenario)
tier_info = classify_tier(state)
kpis = summarize_kpis(state)

now = _dt.datetime.now(_dt.timezone.utc).isoformat()
run_doc = {
    'id': run_id, 'runId': run_id,
    'scenarioId': scenario['scenarioId'],
    'requestedBy': 'crisis.manager',
    'requestedAt': now, 'completedAt': now,
    'status': 'succeeded',
    'tier': tier_info['tier'], 'tierReasons': tier_info['reasons'],
    'rulesVersion': tier_info['rulesVersion'],
    'kpis': kpis, 'resultRef': f'DC-SIM-RESULT/{run_id}',
    'state': state, 'baseline': baseline, 'scenarioInput': scenario
}
print('tier:', run_doc['tier'], '(expected 2 for RSV surge)')
print('peakUtilization:', kpis['peakUtilization'])
print('totalShortfall:', kpis['totalShortfall'])
print('reasons:', tier_info['reasons'])
assert run_doc['tier'] == 2, f'expected tier=2, got tier={run_doc["tier"]}'
print()
print('OK - RSV canonical simulation returned Tier 2 (Besondere Lage)')
```

```python
# Persist the run to Cosmos simulation-runs (via MPE).
runs = db.get_container_client('simulation-runs')
runs.upsert_item(run_doc)
print(f'Wrote run to Cosmos simulation-runs: {run_doc["runId"]}')

retrieved = runs.read_item(item=run_doc['runId'], partition_key=run_doc['runId'])
print(f'Verified read-back: tier={retrieved["tier"]}, rulesVersion={retrieved["rulesVersion"]}')
```
